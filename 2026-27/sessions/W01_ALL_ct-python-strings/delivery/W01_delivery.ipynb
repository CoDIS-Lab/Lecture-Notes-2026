### Why computational thinking?



Take away "data science" from *computational data science* and three things are left: **computation**, **programming** and **algorithms**. They all rest on one skill: turning a messy problem into clear steps.



*Image: [Learning.com](https://www.learning.com/blog/decomposition-in-computational-thinking/)*

> **Computational thinking** is a problem-solving approach that helps us break down complex problems into simpler, more manageable parts.

### The four components



*Image: [TERC blog](https://blog.terc.edu/is-data-science-part-of-computational-thinking)*

* **Decomposition:** breaking a big problem into smaller pieces.
* **Pattern recognition:** identifying similarities and trends.
* **Abstraction:** simplifying complex ideas or data.
* **Algorithm design:** creating step-by-step solutions.

**In data science:**
* **Decomposition:** "clean this dataset" becomes smaller tasks: handle missing values, detect outliers, standardise formats.
* **Abstraction:** a cleaning step you use often (for example, filling in missing values) becomes a reusable procedure you can apply to any dataset.
* **Pattern recognition:** in finance, spotting recurring patterns such as price movements before company earnings reports.
* **Algorithms:** Netflix or Amazon recommendations come from algorithms that analyse what users watch and buy.

### Algorithms



*Image: Ed Harrington, via [INSTRKTIV](https://instrktiv.com/en/ikea-manual/)*

Recipes and flat-pack furniture manuals are algorithms you already follow, and as the picture shows, the order of the steps matters.

> **Algorithmic thinking** is about designing step-by-step instructions to solve a problem or perform a task.

Its key components:
* **Sequencing:** putting steps in the right order.
* **Selection (if–else):** making decisions based on conditions.
* **Repetition (loops):** doing things repeatedly.
* **Iterative refinement:** improving and optimising solutions.

You will write selection and repetition in Python in Week 3.

**A little history.** The word *algorithm* comes from *algorismus*, the Latin form of the name of the Persian mathematician **al-Khwarizmi**, blended with the Greek *arithmos*, "number". **Ada Lovelace**, an English mathematician who worked on Charles Babbage's Analytical Engine, realised that a machine could follow a series of simple instructions (a program) to carry out a complex calculation. She is often called the first computer programmer.

### Pseudocode

**Pseudocode** expresses an algorithm in a simplified, human-readable mix of natural language and simple programming. It:
* is how computer scientists sketch an algorithm;
* is the step between a problem description and actual code;
* is not tied to any programming language.



*Image: ProgrammerHumor.io*

Python's syntax is famously close to pseudocode. You will see how close in Hour 4.

**Example:** pseudocode for the average of three numbers.

```text
Step 1: Start
Step 2: Input num1, num2, num3          ask the user for three numbers
Step 3: Set sum = num1 + num2 + num3    add them up
Step 4: Set average = sum / 3           divide by how many there are
Step 5: Output average                  show the result
Step 6: End
```

**Quick check:** which steps change if you want the average of **four** numbers?

<details><summary>Answer</summary>

Step 2 (input a fourth number), Step 3 (add it) and Step 4 (divide by 4). Notice how decomposition makes the change easy to find.

</details>

### Flowcharts

A **flowchart** is a visual representation of a process or algorithm. It uses standard symbols for the steps, decisions and actions, in order. Flowcharts make complex processes easier to understand, analyse and improve.



You only need four symbols for now: **Terminal** (start/end), **Process** (do something), **Input/Data** (get or show data, the parallelogram) and **Decision** (a yes/no question, the diamond).



*Image: Jorge Cham, [PHD Comics](https://www.phdcomics.com)*

**Try it:** follow the coffee flowchart. Which boxes are **decisions**? Find the path that sends you back to make coffee after all.

A name can be **re-bound** to a different value, even one of a different type, at any time. This is called **dynamic typing**:

| `a = 5` | `b = 5`, then `b = "Hello"` |
|---|---|
|  |  |

### Operators and the order of operations

**Operators** are symbols for computations (`+`, `*`, …); **operands** are the values they act on.



| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` `-` `*` | add, subtract, multiply | `11 * 4` | `44` |
| `/` | division (always gives a `float`) | `11 / 4` | `2.75` |
| `//` | integer (floor) division | `11 // 4` | `2` |
| `%` | modulus (remainder) | `11 % 4` | `3` |
| `**` | power | `2 ** 3` | `8` |

**Augmented assignment** calculates and assigns in one go: `x += 10` means `x = x + 10` (likewise `-=`, `*=`, `/=`, `%=`).

````python
x = 51
x += 10
print(x, 11 / 4, 11 // 4, 11 % 4)
````

Python follows the **PEMDAS** order of operations:
* **P**arentheses;
* **E**xponentiation (`**`);
* **M**ultiplication and **D**ivision (`*`, `/`, `//`, `%`);
* **A**ddition and **S**ubtraction.

Operations with the same precedence are evaluated from left to right.

**Predict:** what is `2 + 3 * 4 ** 2`? And `(2 + 3) * 4`?

````python
print(2 + 3 * 4 ** 2)
print((2 + 3) * 4)
````

`4 ** 2` comes first (16), then `3 * 16` (48), then `2 + 48`, which gives **50**. Parentheses change the order: `(2 + 3) * 4` is **20**.

### Calculator 2.0: build it together

We improve Calculator 1.0 one change at a time. Each change answers a question:

| Change | Why |
|---|---|
| Convert with `float()` **as soon as** the numbers are read | Decimals work, and `a` and `b` are numbers from then on |
| Clean the name with `.strip().title()` | `"  sam "` becomes `"Sam"` |
| Use f-strings instead of `.format()` | Easier to read: each value sits where it appears |
| Show all four operations | More useful, and only one line each |

The cell below starts as a copy of Calculator 1.0. Make the changes there, one at a time, running after each.

````python
# ✍️ Calculator 2.0: start from 1.0 and change it step by step
name = input('Hi, what is your name? ')
print('Hello', name)

print("Let's do a calculation.")
a = input('Enter your first operand: ')
b = input('Enter your second operand: ')

print("Here is the answer ...", "{0} times {1} is {2}".format(a, b, int(a)*int(b)))
print("I hope you have enjoyed it, bye.")
````

### Week 1 takeaways

* **Computational thinking** (decomposition, patterns, abstraction, algorithms) comes before code; pseudocode and flowcharts help you plan.
* Python values have **types**; `"10"` is not `10`, and `input()` always gives a string, so convert with `int()` / `float()`.
* Strings: quotes and escapes, `+` `*` `len()`, **indexing from 0**, **slicing with the end excluded**, and methods such as `.strip()`, `.title()`, `.split()`, `.isdigit()`.
* **f-strings** are the clearest way to put values into text.
* Operators follow **PEMDAS**; `True` counts as 1; `None` means "no value" and is falsy (but not equal to `False`).

**More escape sequences.** `\'` gives a single quote inside a single-quoted string, `\r` is a carriage return, and `\u00e9` writes a Unicode character (é).

**The older `.format()` method.** You will meet it in older code (Calculator 1.0 uses it):
```python
"Happy {}, {}".format("New Year", "Tom")                      # positional
"Happy {0}, {1}".format("New Year", "Tom")                    # explicit index
"Happy {season}, {name}".format(season="New Year", name="Tom")  # named
```
f-strings and `.format()` produce the same result; we use f-strings from now on.

**`print()` options.** `sep` sets what goes between values (default: a space) and `end` what goes at the end (default: a newline `\n`):

````python
print("Eagles", "fly", "high!", sep="..")
print("Hello", end="<<")
print("Class!")
````
