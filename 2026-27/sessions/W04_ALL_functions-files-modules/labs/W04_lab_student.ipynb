Why the `ask` parameter? By default it is the built-in `input`, so the real game asks the player. But a test can pass a **different function** (functions are objects!). In the checkpoint, `lambda prompt: answers.pop(0)` hands out answers from a list, one per call, like a player typing them.

```python
# Checkpoint 2
assert get_guess(ask=lambda prompt: "42") == 42, "get_guess should return the typed number as an int"
answers = ["ten", "4.5", "17"]                 # two typing mistakes, then a whole number
assert get_guess(ask=lambda prompt: answers.pop(0)) == 17, "keep asking until the text is a whole number"
assert answers == [], "get_guess should have asked exactly three times"
print("✔ Task 2")
```

```python
# If your result does not match the checkpoint, run this cell to continue.
def give_hint(secret, guess):
    """Compare guess with secret and return the hint for the player."""
    if guess > secret:
        return "Lower ..."
    elif guess < secret:
        return "Higher ..."
    return "Correct!"

def get_guess(prompt="Take a guess: ", ask=input):
    """Ask until the player types a whole number; return it as an int."""
    text = ask(prompt)
    while not text.strip().isdigit():
        text = ask("Please type a whole number: ")
    return int(text)
```
